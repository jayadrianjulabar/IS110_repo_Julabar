```python
import pandas as pd
import plotly.express as px

# Helper function to convert time strings to a numerical midpoint (in hours)
def parse_time_to_midpoint(time_str):
    start_str, end_str = time_str.split('-')

    def convert_to_24h(t_str):
        t_str = t_str.strip().lower()
        if 'am' in t_str:
            hour = int(t_str.replace('am', ''))
            return hour if hour != 12 else 0  # 12 AM is 0 hours
        elif 'pm' in t_str:
            hour = int(t_str.replace('pm', ''))
            return hour + 12 if hour != 12 else 12 # 12 PM is 12 hours
        else:
            # Handle cases like '12PM' directly if not covered
            if '12pm' in t_str: return 12
            if '12am' in t_str: return 0
            return int(t_str)

    start_hour = convert_to_24h(start_str)
    end_hour = convert_to_24h(end_str)

    # Handle cases where end hour is next day (e.g., 9PM-2AM)
    if end_hour < start_hour:
        end_hour += 24

    return (start_hour + end_hour) / 2

# Original data (assumed to be Jaro Branch)
jaro_hours = pd.DataFrame({
    "Time": ["6AM-9AM", "10AM-12PM", "12PM-3PM", "4PM-10PM"],
    "Customers": [30, 20, 15, 95],
    "Branch": "Jaro"
})

jaro_hours['MidTime'] = jaro_hours['Time'].apply(parse_time_to_midpoint)

# New data for B-Complex Branch (using midpoints for customer ranges)
bcomplex_hours = pd.DataFrame({
    "Time": ["7AM-12PM", "1PM-4PM", "5PM-9PM"],
    "Customers": [25, (5 + 10) / 2, (30 + 40) / 2],
    "Branch": "B-Complex"
})

bcomplex_hours['MidTime'] = bcomplex_hours['Time'].apply(parse_time_to_midpoint)

# Combine the dataframes
combined_hours = pd.concat([jaro_hours, bcomplex_hours])

# Sort by MidTime to ensure lines are drawn correctly
combined_hours = combined_hours.sort_values(by=['MidTime', 'Branch']).reset_index(drop=True)

fig = px.line(
    combined_hours,
    x="MidTime",
    y="Customers",
    title="Peak Active Hours Comparison: Jaro vs B-Complex Branches",
    markers=True,
    color="Branch", # Differentiate lines by branch
    hover_data={'Time': True, 'MidTime': False} # Show original time string on hover
)

# Customize x-axis to show original time labels or a more readable format if needed
fig.update_layout(
    xaxis = dict(
        tickmode = 'array',
        tickvals = combined_hours['MidTime'].unique(),
        ticktext = [f'{int(t) if t%1==0 else t:.1f} ({label})' for t, label in zip(combined_hours['MidTime'].unique(), combined_hours.groupby('MidTime')['Time'].first())],
        title = 'Time (Midpoint in Hours)'
    )
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from dash import Dash, dcc, html, Input, Output
from google.colab import output
import threading

# --- Helper function ---
def parse_time_to_midpoint(time_str):
    start_str, end_str = time_str.split('-')
    def convert_to_24h(t_str):
        t_str = t_str.strip().lower()
        if 'am' in t_str: hour = int(t_str.replace('am', '')); return hour if hour != 12 else 0
        elif 'pm' in t_str: hour = int(t_str.replace('pm', '')); return hour + 12 if hour != 12 else 12
        else:
            if '12pm' in t_str: return 12
            if '12am' in t_str: return 0
            return int(t_str)
    s, e = convert_to_24h(start_str), convert_to_24h(end_str)
    if e < s: e += 24
    return (s + e) / 2

# --- DATASETS and CHARTS ---
combined_walkins = pd.concat([pd.DataFrame({"Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], "Walkins": [70, 72, 69, 28, 71, 75, 73], "Branch": "Jaro"}), pd.DataFrame({"Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], "Walkins": [10]*7, "Branch": "B-Complex"})])
fig_walkins = px.line(combined_walkins, x="Day", y="Walkins", title="Daily Walk-ins Comparison", markers=True, color="Branch")

fig_new_memberships = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "NewMemberships": [88, 100]}), x="Branch", y="NewMemberships", title="Monthly New Memberships", text_auto=True)
fig_active_memberships = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "ActiveMembers": [260, 40]}), x="Branch", y="ActiveMembers", title="Monthly Active Memberships", color="Branch")
fig_new_customers = px.pie(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Customers": [38, 40]}), names="Branch", values="Customers", title="Daily New Customers")

jaro_hours = pd.DataFrame({"Time": ["6AM-9AM", "10AM-12PM", "12PM-3PM", "4PM-10PM"], "Customers": [30, 20, 15, 95], "Branch": "Jaro"})
bcomplex_hours = pd.DataFrame({"Time": ["7AM-12PM", "1PM-4PM", "5PM-9PM"], "Customers": [25, 7.5, 35.0], "Branch": "B-Complex"})
for df in [jaro_hours, bcomplex_hours]: df['MidTime'] = df['Time'].apply(parse_time_to_midpoint)
fig_peak_hours = px.line(pd.concat([jaro_hours, bcomplex_hours]).sort_values('MidTime'), x="MidTime", y="Customers", color="Branch", markers=True, title="Peak Active Hours")

fig_turnover_rate = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "TurnoverRate": [50, 10]}), x="Branch", y="TurnoverRate", title="Monthly Turnover Rate", color="Branch")
fig_maintenance_response = px.line(pd.DataFrame({"Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"], "Hours": [2]*6}), x="Month", y="Hours", title="Equipment Maintenance Response Time", markers=True)
fig_quality_rating = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Rating": [7, 7]}), x="Branch", y="Rating", title="Equipment Quality Rating", text_auto=True).update_yaxes(range=[0,10])
fig_hygiene_percent = go.Figure(go.Indicator(mode="gauge+number", value=70, title={'text': "Equipment Hygiene %"}, gauge={'axis': {'range': [0,100]}}))
fig_injury_incidents = px.bar(pd.DataFrame({"Quarter": ["Q1", "Q2", "Q3", "Q4"], "Incidents": [5, 4, 6, 5]}), x="Quarter", y="Incidents", title="Quarterly Injury Incidents")

chart_figures = {
    "Daily Walk-ins Comparison": fig_walkins, "Monthly New Memberships": fig_new_memberships,
    "Monthly Active Memberships": fig_active_memberships, "Daily New Customers": fig_new_customers,
    "Peak Active Hours Comparison": fig_peak_hours, "Monthly Customer Turnover Rate": fig_turnover_rate,
    "Equipment Maintenance Response Time": fig_maintenance_response, "Equipment Quality Rating": fig_quality_rating,
    "Equipment Hygiene %": fig_hygiene_percent, "Quarterly Injury Incidents": fig_injury_incidents,
}

# --- Dash App Setup ---
app = Dash(__name__)
app.layout = html.Div([
    html.H1("Gym KPI Dashboard", style={'textAlign': 'center'}),
    dcc.Dropdown(id='chart-dropdown', options=[{'label': k, 'value': k} for k in chart_figures.keys()], value="Daily Walk-ins Comparison", clearable=False),
    dcc.Graph(id='display-chart')
])

@app.callback(Output('display-chart', 'figure'), Input('chart-dropdown', 'value'))
def update_graph(name): return chart_figures[name]

# --- RELIABLE FIX: Run on port 8050 and serve as iframe ---
def run_dash():
    app.run(port=8050, debug=False, use_reloader=False)

# Start Dash in a background thread so the cell finishes and allows iframe display
threading.Thread(target=run_dash, daemon=True).start()

# Serve the local port 8050 as an iframe inside the notebook output
output.serve_kernel_port_as_iframe(8050, height=1200)
```

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from dash import Dash, dcc, html
from google.colab import output
import threading

# Clear previous outputs
output.clear()

# --- Helper function ---
def parse_time_to_midpoint(time_str):
    start_str, end_str = time_str.split('-')
    def convert_to_24h(t_str):
        t_str = t_str.strip().lower()
        if 'am' in t_str: hour = int(t_str.replace('am', '')); return hour if hour != 12 else 0
        elif 'pm' in t_str: hour = int(t_str.replace('pm', '')); return hour + 12 if hour != 12 else 12
        else: return int(t_str.replace('12pm','12').replace('12am','0'))
    s, e = convert_to_24h(start_str), convert_to_24h(end_str)
    if e < s: e += 24
    return (s + e) / 2

# --- Defining Figures (Compact sizing) ---
config = {'displayModeBar': False}
layout_settings = dict(margin=dict(l=20, r=20, t=30, b=20), height=250, font=dict(size=10))

fig1 = px.line(pd.concat([pd.DataFrame({"Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], "Walkins": [70, 72, 69, 28, 71, 75, 73], "Branch": "Jaro"}), pd.DataFrame({"Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], "Walkins": [10]*7, "Branch": "B-Complex"})]), x="Day", y="Walkins", title="Daily Walk-ins", markers=True, color="Branch")
fig2 = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Count": [88, 100]}), x="Branch", y="Count", title="New Memberships", text_auto=True)
fig3 = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Active": [260, 40]}), x="Branch", y="Active", title="Active Memberships", color="Branch")
fig4 = px.pie(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Customers": [38, 40]}), names="Branch", values="Customers", title="New Customers")
jaro_h = pd.DataFrame({"Time": ["6AM-9AM", "10AM-12PM", "12PM-3PM", "4PM-10PM"], "Cust": [30, 20, 15, 95], "Branch": "Jaro"})
bcomp_h = pd.DataFrame({"Time": ["7AM-12PM", "1PM-4PM", "5PM-9PM"], "Cust": [25, 7.5, 35], "Branch": "B-Complex"})
for d in [jaro_h, bcomp_h]: d['Mid'] = d['Time'].apply(parse_time_to_midpoint)
fig5 = px.line(pd.concat([jaro_h, bcomp_h]).sort_values('Mid'), x="Mid", y="Cust", color="Branch", markers=True, title="Peak Hours")
fig6 = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Rate": [50, 10]}), x="Branch", y="Rate", title="Turnover Rate", color="Branch")
fig7 = px.line(pd.DataFrame({"Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"], "Hrs": [2]*6}), x="Month", y="Hrs", title="Maint. Resp (Hrs)", markers=True)
fig8 = px.bar(pd.DataFrame({"Branch": ["Jaro", "B-Complex"], "Rating": [7, 7]}), x="Branch", y="Rating", title="Equip Quality").update_yaxes(range=[0,10])
fig9 = go.Figure(go.Indicator(mode="gauge+number", value=70, title={'text': "Hygiene %", 'font': {'size': 12}}, gauge={'axis': {'range': [0,100]}}))
fig10 = px.bar(pd.DataFrame({"Qtr": ["Q1", "Q2", "Q3", "Q4"], "Inc": [5, 4, 6, 5]}), x="Qtr", y="Inc", title="Injury Incidents")

all_charts = [fig1, fig2, fig3, fig4, fig5, fig6, fig7, fig8, fig9, fig10]
for f in all_charts: f.update_layout(**layout_settings)

# --- Dash App (Compact View) ---
app = Dash(__name__)
app.layout = html.Div([
    html.H3("Flex Gym KPI Dashboard", style={'textAlign': 'center', 'margin': '5px', 'fontSize': '18px'}),
    html.Div([
        html.Div([dcc.Graph(figure=f, config=config)],
                 style={'width': '19%', 'display': 'inline-block', 'padding': '2px'})
        for f in all_charts
    ], style={'display': 'flex', 'flex-wrap': 'wrap', 'justify-content': 'center'})
], style={'overflow': 'hidden', 'height': '600px'})

def run_dash():
    app.run(port=8052, debug=False, use_reloader=False)

threading.Thread(target=run_dash, daemon=True).start()
output.serve_kernel_port_as_iframe(8052, height=600)
```
